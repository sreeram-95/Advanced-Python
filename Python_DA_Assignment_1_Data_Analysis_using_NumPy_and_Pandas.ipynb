**Numpy Array Operations:**

```python
# 1. Create a 1D NumPy Array:

import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
```

```python
# 2. Inspection and Properties:

shape = temperatures_w1.shape
dtype = temperatures_w1.dtype
size = temperatures_w1.size

print("Array:", temperatures_w1)
print("Shape:",shape)
print("Data Type:", dtype)
print("Number of Elements:", size)
```

```python
# 3. Array Operations:

   # Convert the temperatures_w1 from Celsius to Fahrenheit
fahrenheit_temperatures = (temperatures_w1 * 9/5) + 32
print("Fahrenheit Temperatures:", fahrenheit_temperatures)
```

```python
# Find the maximum, minimum, and mean temperatures for the week.

max_temp = np.max(temperatures_w1)
min_temp = np.min(temperatures_w1)
mean_temp = np.mean(temperatures_w1)

print("Maximum Temperature:", max_temp)
print("Minimum Temperature:", min_temp)
print("Mean Temperature:", mean_temp)
```

```python
# 4. Array Slicing and Indexing:

# First three days
first_three_days = temperatures_w1[:3]
print("First Three Days:", first_three_days)

# Weekend / Last two days
weekend_days = temperatures_w1[-2:]
print("Weekend / Last Two Days:", weekend_days)

# Middle three days
middle_three_days = temperatures_w1[2:5]
print("Middle Three Days:", middle_three_days)
```

```python
# 5. Create a 2D Array:

week_1 = [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9]
week_2 = [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]

temperatures = np.array([week_1, week_2])
print("Temperatures:", temperatures)
```

```python
# 6. Inspect and Slice the 2D Array:

shape = temperatures.shape
dtype = temperatures.dtype
size = temperatures.size

print("Shape:", shape)
print("Data Type:", dtype)
print("Total Number of Elements:", size)
```

**Pandas Series:**

```python
# 1. Creating Pandas Series:
import pandas as pd

marks = pd.Series(data=[95, 92, 89, 85, 80],
                  index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'],
                  name="Marks")
print(marks)
```

```python
# 2. Indexing and Slicing:

# 1. Integer index position for 1st rank student
first_rank_mark = marks.iloc[0]
print("1st Rank :", first_rank_mark)

# 2. Use .loc to retrieve top 3 ranks by specifying their index labels
top_3_ranks = marks.loc['Rank1':'Rank3']
print("Top 3 Ranks:", top_3_ranks)

# 3. Use .iloc to retrieve mark of 3rd rank student
third_rank_mark = marks.iloc[2]
print("3rd Rank Mark:", third_rank_mark)

# 4. Apply a boolean mask to filter marks > 90
high_marks = marks[marks > 90]
print("Marks > 90:", high_marks)
```

```python
# 3. Manipulating Series:

# Modify the mark of the student with the 1st rank to 100.
marks.loc['Rank1'] = 100
print("Updated Marks:", marks)
```

```python
# Remove the entry corresponding to the last rank from the Series.
marks = marks.drop('Rank5')

print(marks)
```

```python
# Compute the CGPA by dividing each mark in the Series by 10
cgpa = marks / 10
print("CGPA:", cgpa)
```

**Pandas DataFrame:**

```python
# 1. Creating Pandas DataFrame:

data = {'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing','Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West','North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]}

transactions = pd.DataFrame(data)
print(transactions)
```

```python
# 2. Data Exploration:

# Display the transactions DataFrame and its basic information, including head, tail, shape, column names, and data types.

print(transactions.head())
print(transactions.tail())
print(transactions.shape)
print(transactions.columns)
transactions.info()
```

```python
# Display only the 'ProductCategory' and 'Amount' columns.

category_amount = transactions[['ProductCategory', 'Amount']]
print(category_amount)
```

```python
#  Retrieve the last 3 columns using iloc

last_3_cols = transactions.iloc[:, -3:]
print(last_3_cols)
```

```python
# Filter rows: Region == 'North' and Amount > 200

filtered_north = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
print(filtered_north)
```

```python
#  Value counts for 'ProductCategory'

category_counts = transactions['ProductCategory'].value_counts()
print(category_counts)
```

```python
# Unique values in 'Region'
unique_regions = transactions['Region'].unique()
print(unique_regions)
```

```python
#  Group by 'Region' and calculate mean Amount
mean_amount_by_region = transactions.groupby('Region')['Amount'].mean()
print(mean_amount_by_region)
```

```python
# 3. Manipulating the DataFrame:

# Modify 'Amount' for TransactionID 102 to 165
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print(transactions)
```

```python
# Add a new column 'Discount' (10% of 'Amount')

transactions['Discount'] = transactions['Amount'] * 0.10
print(transactions)
```

```python
# Remove the row with TransactionID 109

transactions.drop(transactions[transactions['TransactionID'] == 109].index, inplace=True)
print(transactions)
```

```python
# Delete the 'Discount' column from the DataFrame

transactions = transactions.drop(columns=['Discount'])
print(transactions)
```
